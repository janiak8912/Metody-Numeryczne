## Błąd względny i bezwzględny

***Zadanie 1.***  
Liczbę $\textrm{e}$ możemy zdefiniować jako $\sum\limits_{n=0}^{\infty} \frac{1}{n!}$. Oblicz błąd względny i bezwzględny aproksymacji liczby $\textrm{e}$ poprzez $\textrm{e}^{*}$ w przypadku gdy:  
* $\textrm{e}_1^{*}=\sum\limits_{n=0}^{5} \frac{1}{n!}$    

* $\textrm{e}_2^{*}=\sum\limits_{n=0}^{10} \frac{1}{n!}$  

* $\textrm{e}_3^{*}=\sum\limits_{n=0}^{20} \frac{1}{n!}$  

```python
import math
from decimal import Decimal, getcontext

getcontext().prec = 50

def factorial(k):
    if k == 0: return 1
    res = 1
    for i in range(1, k + 1):
        res *= i
    return res

def calculate_e_star(n_limit):
    return sum(Decimal(1) / Decimal(factorial(n)) for n in range(n_limit + 1))

e_exact = Decimal('2.71828182845904523536028747135266249775724709369996')

n_values = [5, 10, 20]

print(f"{'n':>2} | {'Aproksymacja e*':>25} | {'Błąd bezwzględny':>20} | {'Błąd względny [%]':>20}")
print("-" * 95)

for n in n_values:
    e_star = calculate_e_star(n)
    
    err_abs = abs(e_exact - e_star)
    err_rel = (err_abs / e_exact) * 100
    
    print(f"{n:2d} | {e_star:25.20f} | {err_abs:20.2e} | {err_rel:20.2e}%")
```
